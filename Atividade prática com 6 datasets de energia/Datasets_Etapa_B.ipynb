# **Dataset 1 — Appliances Energy Prediction**

**Situação:** identificar períodos de consumo elevado de eletrodomésticos e observar as condições de temperatura e umidade nesses momentos.

### 1. Carregar a amostra e inspecionar

```python
import pandas as pd

# Troque pelo nome do arquivo exportado do Orange (Save Data)
df1 = pd.read_csv('/content/Amostra_base_Appliances_Energy_Prediction.csv')

df1.head()
```

### 2. Renomear colunas

Renomeamos a coluna `Appliances` para `Consumo_Eletrodomesticos` e simplificamos pelo menos três atributos ambientais.

```python
df1.rename(columns={
    'Appliances': 'Consumo_Eletrodomesticos',
    'lights': 'Luzes',
    'T1': 'Temp_Cozinha',
    'RH_1': 'Umid_Cozinha',
    'T2': 'Temp_Sala',
    'RH_2': 'Umid_Sala',
    'T3': 'Temp_Lavanderia',
    'RH_3': 'Umid_Lavanderia',
}, inplace=True)

df1.columns
```

### 3. Maior consumo de eletrodomésticos registrado

```python
consumo_max1 = df1['Consumo_Eletrodomesticos'].max()
consumo_max1
```

### 4. Limiar de 70% do máximo e DataFrame de consumo elevado

```python
limiar1 = 0.70 * consumo_max1
df1_alto_consumo = df1[df1['Consumo_Eletrodomesticos'] > limiar1]
limiar1
```

### 5. Contagem e percentual da amostra

```python
qtd1 = df1_alto_consumo.shape[0]
pct1 = qtd1 / df1.shape[0] * 100
print(f'Registros acima do limiar: {qtd1}')
print(f'Percentual da amostra: {pct1:.2f}%')
```

### 6. Temperatura média (T1) e segundo DataFrame (consumo alto E temperatura acima da média)

```python
temp_media1 = df1['Temp_Cozinha'].mean()

df1_alto_consumo_e_temp = df1[
    (df1['Consumo_Eletrodomesticos'] > limiar1) &
    (df1['Temp_Cozinha'] > temp_media1)
]

print(f'Temperatura média (Temp_Cozinha): {temp_media1:.2f}')
print(f'Registros com consumo alto E temperatura acima da média: {df1_alto_consumo_e_temp.shape[0]}')
```

### 7. Comparação entre os dois DataFrames

A maior parte dos picos de consumo de energia dos eletrodomésticos (64% dos casos analisados) ocorreu sem que a cozinha estivesse com temperaturas acima da média. Isso indica que o alto consumo nesta residência não é provocado pelo calor do ambiente, mas provavelmente por outros fatores, como hábitos dos moradores, horários específicos ou o uso de vários aparelhos ao mesmo tempo.

Ademais, ao aplicar dois filtros simultâneos (alto consumo + alta temperatura), o número de registros diminuiu drasticamente (de 22 para apenas 8), confirmando que a adição de regras reduz ou mantém o tamanho da amostra, jamais o aumenta.

```python
print(f'Somente consumo alto: {df1_alto_consumo.shape[0]} registros')
print(f'Consumo alto + temperatura acima da média: {df1_alto_consumo_e_temp.shape[0]} registros')
```


# **Dataset 2 — Steel Industry Energy Consumption**

**Situação:** localizar cenários de consumo elevado na indústria siderúrgica e verificar se coincidem com carga elevada ou fator de potência desfavorável.

### 1. Carregar a amostra e renomear colunas

```python
df2 = pd.read_csv('/content/Amostra_base_Steel_Industry_Energy_Consumption.csv')

df2 = df2.rename(columns={
    'Usage_kWh': 'Consumo_kWh',
    'Lagging_Current_Reactive.Power_kVarh': 'Pot_Reativa_Atraso',
    'Leading_Current_Reactive_Power_kVarh': 'Pot_Reativa_Adiantada',
    'Lagging_Current_Power_Factor': 'Fator_Potencia_Atraso',
    'Leading_Current_Power_Factor': 'Fator_Potencia_Adiantado',
})

df2.columns
```

### 2. Inspeção inicial

```python
df2.head()
```

```python
#df2.shape
print(f'Total de registros: {df2.shape[0]}')
print(f'Total de colunas: {df2.shape[1]}')
```

```python
df2.info()
```

```python
df2.describe()
```

### 3. Maior consumo registrado e limiar de 75%

```python
consumo_max2 = df2['Consumo_kWh'].max()
limiar2 = 0.75 * consumo_max2
print(f'Consumo máximo: {consumo_max2}')
print(f'Limiar (75%): {limiar2:.2f}')
```

### 4. DataFrame de consumo elevado, quantidade e percentual

```python
df2_alto_consumo = df2[df2['Consumo_kWh'] > limiar2]

qtd2 = df2_alto_consumo.shape[0]
pct2 = qtd2 / df2.shape[0] * 100
print(f'Registros acima do limiar: {qtd2}')
print(f'Percentual da amostra: {pct2:.2f}%')
```

### 5. Registros que pertencem a 'Maximum Load'

```python
qtd_maximum_load = (df2_alto_consumo['Load_Type'] == 'Maximum_Load').sum()
print(f'Registros de consumo elevado com Load_Type = Maximum_Load: {qtd_maximum_load}')
print(f'Percentual dentro do grupo de consumo elevado: {qtd_maximum_load / qtd2 * 100:.2f}%')
```

### 6. Explorar o fator de potência e definir um limite para valores baixos

```python
df2['Fator_Potencia_Atraso'].describe()
```

Observe o `describe()` acima (mín, quartis, média) e escolha um limite coerente para "fator de potência baixo" — por exemplo, um valor próximo do 1º quartil ou um valor de referência técnico (fatores de potência abaixo de 0.85–0.90 costumam ser considerados baixos). Ajuste o valor abaixo conforme os dados observados.

```python
limite_fator_potencia = 65  # baseado no 1º quartil (63.60) do describe() acima
limite_fator_potencia
```

### 7. DataFrame com consumo elevado E fator de potência abaixo do limite

```python
df2_alto_consumo_fp_baixo = df2[
    (df2['Consumo_kWh'] > limiar2) &
    (df2['Fator_Potencia_Atraso'] < limite_fator_potencia)
]

qtd3 = df2_alto_consumo_fp_baixo.shape[0]
pct3 = qtd3 / df2.shape[0] * 100
print(f'Registros com consumo elevado e fator de potência baixo: {qtd3}')
print(f'Percentual da amostra: {pct3:.2f}%')
```

### 8. Por que esse segundo conjunto merece mais atenção?

Dos 129 registros (1,84%) com consumo acima de 75% do máximo, 62 (48,06%) já eram de carga máxima (Maximum_Load), o que é coerente, já que picos de consumo tendem a coincidir com esse tipo de carga. Ainda assim, nenhum desses registros apresentou fator de potência abaixo de 65 ao mesmo tempo, indicando que, mesmo nos momentos de maior demanda, a eficiência elétrica se manteve em níveis aceitáveis.

Portanto, esse cruzamento continua relevante como indicador de acompanhamento: se em análises futuras ele deixar de ser zero, sinalizará ineficiência elétrica justamente nos momentos mais custosos para a operação, ou seja, quando o impacto financeiro e o desgaste dos equipamentos são maiores.



# **Dataset 3 — Power Consumption of Tetouan City**

**Situação:** identificar qual das três zonas de distribuições de energia em Tetouan apresenta o maior pico de consumo e observar as condições ambientais registradas nos momentos de maior demanda

### 1. Carregar a amostra e inspecionar

```python
url3 = '/content/Amostra_base_Tetouan_City.csv'

df3 = pd.read_csv(url3)

df3.columns = (
    df3.columns
       .str.replace(r'\s+', ' ', regex=True)
       .str.strip()
)

df3.head()
```

```python
print(f'Total de registros: {df3.shape[0]}')
print(f'Total de colunas: {df3.shape[1]}')
```

```python
df3.info()
df3.describe()
```

### 2. Renomear as colunas de consumo

```python
df3 = df3.rename(columns={
    'Zone 1 Power Consumption': 'Consumo_Zona_1',
    'Zone 2 Power Consumption': 'Consumo_Zona_2',
    'Zone 3 Power Consumption': 'Consumo_Zona_3'
})

df3.columns
```

### 3. Consumo máximo registrado em cada zona

```python
maximos3 = df3[
    ['Consumo_Zona_1', 'Consumo_Zona_2', 'Consumo_Zona_3']
].max()

print('Consumo máximo de cada zona:')
print(maximos3)
```

### 4. Identificar a zona com o maior pico de consumo

```python
zona_maior3 = maximos3.idxmax()
maior_pico3 = maximos3.max()

print(f'Zona com o maior pico: {zona_maior3}')
print(f'Maior consumo registrado: {maior_pico3:.2f}')
```

### 5. Limiar de 70% e DataFrame de consumo elevado

```python
limiar3 = 0.70 * maior_pico3

df3_alto_consumo = df3[
    df3[zona_maior3] > limiar3
]

print(f'Limiar de 70%: {limiar3:.2f}')
df3_alto_consumo.head()
```

### 6. Quantidade e percentual de registros com consumo elevado

```python
qtd_alto3 = df3_alto_consumo.shape[0]
pct_alto3 = qtd_alto3 / df3.shape[0] * 100

print(f'Registros acima do limiar: {qtd_alto3}')
print(f'Percentual da amostra: {pct_alto3:.2f}%')
```

### 7. Consumo elevado e temperatura acima da média

```python
temperatura_media3 = df3['Temperature'].mean()

df3_alto_consumo_temp = df3[
    (df3[zona_maior3] > limiar3) &
    (df3['Temperature'] > temperatura_media3)
]

print(f'Temperatura média: {temperatura_media3:.2f}')
print(
    'Registros com consumo elevado e temperatura acima da média:',
    df3_alto_consumo_temp.shape[0]
)
```

### 8. Comparação entre os dois DataFrames

```python
reducao3 = qtd_alto3 - df3_alto_consumo_temp.shape[0]
percentual_reducao3 = reducao3 / qtd_alto3 * 100

print(f'Somente consumo elevado: {qtd_alto3} registros')
print(
    'Consumo elevado e temperatura acima da média:',
    df3_alto_consumo_temp.shape[0],
    'registros'
)
print(f'Redução de registros: {reducao3}')
print(f'Redução percentual: {percentual_reducao3:.2f}%')
```

### 9. Interpretação

A Zona 1 apresentou o maior pico de consumo da amostra, com aproximadamente
51.571,50. Ao considerar como consumo elevado os valores acima de 70% desse
máximo, foram encontrados 2.388 registros, equivalentes a 30,37% da amostra.

Ao incluir também a condição de temperatura acima da média de 18,79, o número
de registros diminuiu para 1.597. Isso representa uma redução de 791 registros,
ou 33,12% do conjunto de consumo elevado.

A redução acontece porque o segundo DataFrame exige que duas condições sejam
verdadeiras simultaneamente: consumo alto na Zona 1 e temperatura acima da
média. O resultado mostra uma relação dentro da amostra analisada, mas não
permite afirmar que a temperatura seja a causa do aumento do consumo.

de cada inversor, os horários de operação e as condições ambientais.

# **Dataset 5 — Wind & Solar Energy Production Dataset**

**Situação:** comparar a frequência de períodos de alta geração solar e de alta
geração eólica, considerando o máximo próprio de cada fonte.

### 1. Carregar a amostra e inspecionar

```python
url5 = '/content/Amostra_base_Wind_Solar_Energy_Production.csv'

df5 = pd.read_csv(url5)

df5.head()
```

```python
print(f'Total de registros: {df5.shape[0]}')
print(f'Total de colunas: {df5.shape[1]}')
```

```python
df5.info()
```

```python
df5.describe()
```

### 2. Verificar as fontes de geração disponíveis

```python
df5['Source'].value_counts()
```

### 3. Separar os dados solar e eólico e renomear as variáveis

```python
df5_solar = df5[df5['Source'] == 'Solar'].copy()
df5_eolica = df5[df5['Source'] == 'Wind'].copy()

df5_solar = df5_solar.rename(
    columns={'Production': 'Geracao_Solar'}
)

df5_eolica = df5_eolica.rename(
    columns={'Production': 'Geracao_Eolica'}
)

print(f'Registros solares: {df5_solar.shape[0]}')
print(f'Registros eólicos: {df5_eolica.shape[0]}')
```

### 4. Valor máximo de geração para cada fonte

```python
max_solar5 = df5_solar['Geracao_Solar'].max()
max_eolica5 = df5_eolica['Geracao_Eolica'].max()

print(f'Máximo da geração solar: {max_solar5}')
print(f'Máximo da geração eólica: {max_eolica5}')
```

### 5. Limiares de 70% do máximo de cada fonte

```python
limiar_solar5 = 0.70 * max_solar5
limiar_eolica5 = 0.70 * max_eolica5

print(f'Limiar solar (70%): {limiar_solar5:.2f}')
print(f'Limiar eólico (70%): {limiar_eolica5:.2f}')
```

### 6. DataFrames de alta geração solar e eólica

```python
df5_solar_alta = df5_solar[
    df5_solar['Geracao_Solar'] > limiar_solar5
]

df5_eolica_alta = df5_eolica[
    df5_eolica['Geracao_Eolica'] > limiar_eolica5
]

print('Exemplo de registros de alta geração solar:')
display(df5_solar_alta.head())

print('Exemplo de registros de alta geração eólica:')
display(df5_eolica_alta.head())
```

### 7. Quantidade e percentual de registros de alta geração

```python
qtd_solar_alta5 = df5_solar_alta.shape[0]
qtd_eolica_alta5 = df5_eolica_alta.shape[0]

pct_solar_alta5 = (
    qtd_solar_alta5 / df5_solar.shape[0] * 100
)

pct_eolica_alta5 = (
    qtd_eolica_alta5 / df5_eolica.shape[0] * 100
)

print(f'Registros de alta geração solar: {qtd_solar_alta5}')
print(f'Percentual dentro da geração solar: {pct_solar_alta5:.2f}%')

print(f'Registros de alta geração eólica: {qtd_eolica_alta5}')
print(f'Percentual dentro da geração eólica: {pct_eolica_alta5:.2f}%')
```

### 8. Comparação entre as fontes

```python
if pct_solar_alta5 > pct_eolica_alta5:
    print('A geração solar aparece com maior frequência acima de 70% do seu próprio máximo.')
elif pct_eolica_alta5 > pct_solar_alta5:
    print('A geração eólica aparece com maior frequência acima de 70% do seu próprio máximo.')
else:
    print('As duas fontes aparecem com a mesma frequência acima de 70% do máximo.')
```

### 9. Interpretação

Na amostra analisada, a geração solar apresentou valor máximo de 16.316,
enquanto a geração eólica apresentou máximo de 22.929. Por possuírem escalas
diferentes, foram calculados limiares próprios de 70%: 11.421,20 para Solar e
16.050,30 para Wind.

Foram encontrados 45 registros de alta geração solar, equivalentes a 2,44%
dos registros solares. Para a geração eólica, foram encontrados 245 registros
de alta geração, equivalentes a 2,87% dos registros eólicos.

Portanto, a geração eólica apareceu com maior frequência acima de 70% do seu
próprio máximo. Não seria adequado aplicar o mesmo limite numérico às duas
fontes, pois a escala máxima da geração eólica é maior que a solar. Um único
limite poderia tornar a comparação injusta, selecionando uma quantidade
desproporcional de registros de uma das fontes.

segunda condição confirma a associação entre alta demanda e corrente elevada,
